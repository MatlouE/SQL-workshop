<h4>Clean Price and percentage columns and convert to numeric</h4>

```python
df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',','').str.strip().astype(float)
```

```python
df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').str.strip().astype(float)
```

```python
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').str.strip().astype(float)
#convert rating column to numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

df['rating_count'] = df['rating_count'].str.replace(',','').str.strip()
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
print(df.dtypes)
```

<h3>Handling Missing data</h3>

```python
df.isnull().sum()
```

```python
#fill null values with the median
df['rating'].fillna(df['rating'].median(), inplace=True)
```

```python
df['rating_count'].fillna(0, inplace=True)
```

Cleaning text columns

```python
text_columns = ['product_name', 'user_name', 'review_title', 'review_content', 'about_product']
for col in text_columns:
    df[col] = df[col].str.strip()
    
for col in text_columns:
    df[col] = df[col].replace('', 'Unknown')
```
